# Interactive Impurity‑Decrease Explorer (Plotly edition)
Drag the slider and watch the impurity decrease update.

```python

import numpy as np, plotly.graph_objects as go, ipywidgets as widgets
from IPython.display import display, HTML

points = np.array([
    [0.10,0.85,1],[0.20,0.70,0],[0.30,0.40,1],[0.40,0.90,0],[0.55,0.35,1],
    [0.60,0.80,0],[0.70,0.20,1],[0.80,0.55,0],[0.90,0.60,0],[0.25,0.15,0]
])
X, Y, labels = points[:,0], points[:,1], points[:,2].astype(int)
def impurity(p0,p1): return 0.5*(p0*(1-p0)+p1*(1-p1))
p0_parent = (labels==0).mean(); p1_parent = 1-p0_parent
parent_imp = impurity(p0_parent,p1_parent)

fig = go.FigureWidget()
fig.add_scatter(x=X[labels==0],y=Y[labels==0],mode='markers',
                marker=dict(color='steelblue',size=12,line=dict(width=1,color='black')),
                name='Class 0')
fig.add_scatter(x=X[labels==1],y=Y[labels==1],mode='markers',
                marker=dict(symbol='square',color='darkorange',size=12,line=dict(width=1,color='black')),
                name='Class 1')
init_split=0.5
fig.update_layout(shapes=[dict(type='line',x0=init_split,x1=init_split,y0=0,y1=1,
                               line=dict(color='red',width=3))],
                  width=450,height=450,
                  xaxis=dict(range=[0,1],title='x'),
                  yaxis=dict(range=[0,1],title='y',scaleanchor='x',scaleratio=1),
                  title='Move slider')

slider=widgets.FloatSlider(value=init_split,min=0.05,max=0.95,step=0.01,
                           description='Split x',readout_format='.2f',
                           style={'description_width':'60px'},layout=widgets.Layout(width='450px'))
out=widgets.HTML()
def update(val):
    x_split=val
    fig.layout.shapes[0].x0=x_split; fig.layout.shapes[0].x1=x_split
    L=X<x_split; R=~L; nL=L.sum(); nR=R.sum()
    if nL==0 or nR==0:
        out.value="<b style='color:red'>Invalid split (empty child)</b>"; return
    p0L=((labels==0)&L).sum()/nL; p0R=((labels==0)&R).sum()/nR
    p1L, p1R = 1-p0L,1-p0R
    iL,iR=impurity(p0L,p1L),impurity(p0R,p1R)
    propL,propR=nL/len(labels),nR/len(labels)
    dec=parent_imp-(propL*iL+propR*iR)
    out.value=f'''
    <style>table{{border-collapse:collapse}} td{{padding:2px 8px}}</style>
    <h4>Split at <span style="color:red;">x={x_split:.2f}</span></h4>
    <table border="1">
    <tr><td><b>Node</b></td><td>n</td><td>p0</td><td>p1</td><td>i</td></tr>
    <tr><td>Parent</td><td>{len(labels)}</td><td>{p0_parent:.2f}</td><td>{p1_parent:.2f}</td><td>{parent_imp:.3f}</td></tr>
    <tr><td>Left</td><td>{nL}</td><td>{p0L:.2f}</td><td>{p1L:.2f}</td><td>{iL:.3f}</td></tr>
    <tr><td>Right</td><td>{nR}</td><td>{p0R:.2f}</td><td>{p1R:.2f}</td><td>{iR:.3f}</td></tr>
    </table>
    <p><b>ΔI = {dec:.3f}</b></p>
    '''
update(init_split)
def on_change(change): update(change['new'])
slider.observe(on_change,names='value')
display(fig,slider,out)
```
