# Locally Lipschitz Cone with Adjustable Slope

In this notebook, we illustrate a **Lipschitz cone** sliding along the graph of \(f(x)=x^2\) with:
- **Slider for the center point** \(x_0\).
- **Slider for the Lipschitz constant** \(L\), which controls the cone's angle.
- The static function graph remains unchanged, and the cone angle adjusts with **L**.


```python
from pathlib import Path

OUT_DIR = Path("out")
OUT_DIR.mkdir(parents=True, exist_ok=True)

def out_path(filename):
    return OUT_DIR / filename
```

```python
import sys
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import FloatSlider, interact

ROOT = Path.cwd()
while not (ROOT / "xai_book").exists() and ROOT != ROOT.parent:
    ROOT = ROOT.parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from xai_book.plotting import BLUE as bookBlue, CHAPTER_GRAY as chaptergray, ORANGE as bookOrange, apply_notebook_style

apply_notebook_style(figsize=(10.0, 6.0))

def f(x):
    return x**2

xs_full = np.linspace(-5, 5, 400)
ys_full = f(xs_full)

def plot_and_save(x0=2.9, L=3.4, save=True):
    line_plus = f(x0) + L * (xs_full - x0)
    line_minus = f(x0) - L * (xs_full - x0)

    plt.figure(figsize=(10, 6))
    plt.plot(xs_full, ys_full, color=bookBlue, linewidth=2, label='$f(x)=x^2$')
    plt.fill_between(
        xs_full,
        line_minus,
        line_plus,
        color=bookOrange,
        alpha=0.35,
        label='Lipschitz cone',
    )
    plt.plot(
        xs_full,
        line_plus,
        '--',
        color=bookOrange,
        label=f'$f(x_0) \pm L(x - x_0)$, L={L:.2f}',
    )
    plt.plot(xs_full, line_minus, '--', color=bookOrange)
    plt.scatter([x0], [f(x0)], color=chaptergray, s=70, label='$x_0$')

    plt.title('Lipschitz Cone over Static $f(x)=x^2$', fontsize=14)
    plt.xlabel('$x$', fontsize=12)
    plt.ylabel('$f(x)$', fontsize=12)
    plt.legend()
    plt.grid(True)
    plt.xlim(-5, 5)
    plt.ylim(0, 25)

    if save:
        output_path = out_path('locally_lipschitz.pdf')
        plt.savefig(output_path, dpi=300, bbox_inches='tight')
        print(f"Plot saved as '{output_path}'")

interact(
    plot_and_save,
    x0=FloatSlider(value=2.9, min=-5.0, max=5.0, step=0.1, description='$x_0$'),
    L=FloatSlider(value=3.4, min=0.5, max=5.0, step=0.1, description='$L$'),
    save=False,
)
```
